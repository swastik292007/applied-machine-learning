#### 14. Create a 2d array with 1 on the border and 0 inside 

````python
import numpy as np

arr = np.zeros((5, 5), dtype=int)

arr[0, :] = 1      # top
arr[-1, :] = 1     # bottom
arr[:, 0] = 1      # left
arr[:, -1] = 1     # right

print(arr)
````

#### 18. Normalize a 5x5 random matrix 

````python
import numpy as np

# Create 5x5 random matrix
arr = np.random.rand(5, 5) * 100

print("Original Matrix:")
print(arr)

# Min-Max normalization
normalized = (arr - arr.min()) / (arr.max() - arr.min())

print("\nNormalized Matrix:")
print(normalized)
````

#### 19. Multiply a 5x3 matrix by a 3x2 matrix (real matrix product) 

````python
import numpy as np

A = np.random.randint(1, 10, (5, 3))
B = np.random.randint(1, 10, (3, 2))

C = np.dot(A, B)

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nResult:")
print(C)
````

#### 20. Given a 1D array, negate all elements which are between 3 and 8, in place. 

````python
import numpy as np

arr = np.array([1, 3, 5, 7, 9, 4, 10, 8])

arr[(arr >= 3) & (arr <= 8)] *= -1

print(arr)
````
