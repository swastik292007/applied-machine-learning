```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([35, 40, 50, 55, 65])

# Calculate slope and intercept
m, c = np.polyfit(x, y, 1)

print("Slope:", m)
print("Intercept:", c)

# Prediction
y_pred = m * x + c
print("Predicted values:", y_pred)
# Error
error = y - y_pred
#MSE
mse = np.mean(error ** 2)

# MAE
mae = np.mean(np.abs(error))

# RMSE
rmse = np.sqrt(mse)

print("MSE:", mse)
print("MAE:", mae)
print("RMSE:", rmse)
```
